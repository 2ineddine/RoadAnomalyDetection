```python
import yaml, os
from ultralytics import YOLO
import kagglehub


src_yaml = "/kaggle/input/datasets/sreekaraditya/rdd2022-yolo-crackscan-v2/data.yaml"
dst_yaml = "/kaggle/working/rdd2022.yaml"
base_dir = "/kaggle/input/datasets/sreekaraditya/rdd2022-yolo-crackscan-v2"

with open(src_yaml) as f:
    cfg = yaml.safe_load(f)

cfg["path"] = base_dir

with open(dst_yaml, "w") as f:
    yaml.dump(cfg, f)

print("Fixed YAML:", cfg)


def upload_checkpoint(trainer):
    weights_dir = "/kaggle/working/runs/finetune/weights"

    kagglehub.dataset_upload(
        "Ziner7/yolo26n",
        weights_dir,
        version_notes=f"Epoch {trainer.epoch + 1}"
    )

    print(f"Checkpoint uploaded: epoch {trainer.epoch + 1}")


model = YOLO("yolo26n.pt")

model.add_callback("on_model_save", upload_checkpoint)

model.train(
    data=dst_yaml,
    epochs=100,
    imgsz=640,
    batch=64,
    device=0,
    project="/kaggle/working/runs",
    name="finetune",
    patience=10,
)


best = "/kaggle/working/runs/finetune/weights/best.pt"
model = YOLO(best)

metrics = model.val(data=dst_yaml)
print(f"mAP50:    {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
```
